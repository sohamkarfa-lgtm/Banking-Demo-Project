```python
# Databricks notebook source
from src.config.config import Config
from src.config.entity_config import EntityConfig
from src.utils.logger import Logger
from src.utils.path import PathUtils
from src.utils.file_mover import FileMover

config = Config()
entity_config = EntityConfig()

logger = Logger("source-file-backup")
path_utils = PathUtils(config)

file_mover = FileMover(
    dbutils=dbutils,
    logger=logger
)

for entity in entity_config.entities:
    source_path = path_utils.raw_path(entity)
    backup_base_path = path_utils.backup_base_path()

    logger.info(f"Starting source backup for entity: {entity}")

    moved_count = file_mover.move_files(
        source_path=source_path,
        backup_base_path=backup_base_path,
        entity=entity
    )

    logger.info(
        f"Completed source backup for entity: {entity}. Files moved: {moved_count}"
    )
```
